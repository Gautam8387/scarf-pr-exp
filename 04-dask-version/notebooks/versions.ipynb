```python
import pandas as pd
import os
import numpy as np
import pickle as pkl

import matplotlib.pyplot as plt
```

```python
def parse_version(version_str):
    version = version_str.split('.')[1:]
    return tuple(map(int, version))
```

```python
results = np.sort(os.listdir('../results'))
results = [x for x in results if not x.endswith('.csv')]
results = sorted(results, key=lambda x: parse_version(x))
results = ['../results/'+x+'/results.pkl' for x in results]
results
```

```python
dask_versions = []
ram_sizes = []
ram_cols = ['baron_8K_pancreas_rnaseq.zarr', 'hca_783K_blood_rnaseq.zarr', 'saunders_110K_brain_rnaseq.zarr', 'tenx_1.3M_brain_rnaseq.zarr', 'zheng_69K_pbmc_rnaseq.zarr']
for result in results:
    with open(result, 'rb') as f:
        res = pkl.load(f)
    dask_versions.append(res['das_version'])
    ram_sizes.append([np.max(res[x]['ram']) for x in ram_cols])
```

```python
df = {'dask_version': dask_versions}
for i, col in enumerate(ram_cols):
    df[col] = [x[i] for x in ram_sizes]
df = pd.DataFrame(df)
df
```

# Plots

```python
for result in results:
    with open(result, 'rb') as f:
        res = pkl.load(f)
    print(res['das_version'])
    os.makedirs('../plots/'+res['das_version'], exist_ok=True)
    for col in ram_cols:
        print(col)
        plt.plot(res[col]['ram'])
        plt.title(col)
        plt.xlabel('Time (s)')
        plt.ylabel('RAM (MB)')
        plt.savefig('../plots/'+res['das_version']+'/'+col+'.png')
        plt.close()
    print()
    print()
```
